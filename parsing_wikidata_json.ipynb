```python
def parse_wikidata(result):
    dict_result = ast.literal_eval(result)['results']['bindings'][0]
    keys = dict_result.keys()
    key = next((k for k in keys if k.endswith('LabelLabel')), None)
    if key ==None:
        key = next((k for k in keys if k.endswith('Label')), None)

    
    name = dict_result[key]['value']
    return name
    
    
    
```
