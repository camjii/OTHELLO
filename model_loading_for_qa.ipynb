```python
from huggingface_hub import login
import os
from dotenv import load_dotenv


os.environ["HF_HOME"] = "./hf_home"  # stores logins

hf_token = os.getenv('HF_TOKEN')
login(token=hf_token)


#mkdir models
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import os


variants = ['base', 'large', 'small']
model_ids = [f"google/flan-t5-{var}" for var in variants]

for model_id in model_ids:
    
    save_path = f"./models/{model_id}" 
    os.makedirs(save_path, exist_ok=True)

    # Download from hub
    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForSeq2SeqLM.from_pretrained(
        model_id,
        dtype="auto", #For weight loading
        device_map="auto"
    )

    # Save tokenizer + model as a single file
    tokenizer.save_pretrained(save_path)
    model.save_pretrained(
        save_path,
        safe_serialization=True, #Using .safetensors
    )

    print("Model and tokenizer saved into:", save_path)
```
